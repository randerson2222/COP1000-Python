```python
# Exercise 1: Create a Greeting
first_name = input("Enter your first name: ")
print(f"Hello, {first_name}! Welcome to Python.")
```

```python
# Exercise 2: Collect Multiple Inputs
first_name = input("Enter your first name: ")
city = input("Enter the name of the city you live in: ")
favorite_food = input("Enter your favorite food: ")
print(f"{first_name} lives in {city} and their favorite food is {favorite_food}.")
```

```python
# Exercise 3: Convert an Integer
age = int(input("Enter your age: "))
print(f"You are currently {age} years old.")
print(f"Next year you will be {age + 1} years old.")
```

```python
# Exercise 4: Add Two Numbers
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
sum = num1 + num2
print(f"The sum of {num1} and {num2} is {sum}.")
```

```python
# Exercise 5: Calculate a Purchase
item_name = input("Enter the item: ")
item_price = float(input("Enter the price: "))
item_qty = int(input("Enter the quantity: "))
total_cost = item_price * item_qty
print("== Purchase Summary ==")
print(f"Item: {item_name}")
print(f"Price: ${item_price:.2f}")
print(f"Quantity: {item_qty}")
print(f"Total Cost: ${total_cost:.2f}")
# == Purchase Summary ==
# Item: Notebook
# Price: $3.49
# Quantity: 4
# Total Cost: $13.96
```

```python
# Exercise 6: Calculate Gross Pay
# ---
# Hours worked: 20
# Hourly rate: 15.50
#
# Gross Pay: $310.00
# ---
# gross pay = hours worked * hourly rate
hours_worked = float(input("Enter the hours worked: "))
hourly_rate = float(input("Enter the hourly rate: "))
gross_pay = hours_worked * hourly_rate
print("")
print(f"Gross Pay: ${gross_pay:.2f}")
```
