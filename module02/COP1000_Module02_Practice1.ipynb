```python
# Exercise 1: Create Variables
first_name = "Richard"
age = 24
temperature_f = 91.2
is_hot_outside = True
print("First Name: " + first_name)
print("Age: " + str(age))
print("Temperature: " + str(temperature_f))
print("Is Hot Outside: " + str(is_hot_outside))
```

```python
# Exercise 2: Identify Data Types
print(type(first_name))
print(type(age))
print(type(temperature_f))
print(type(is_hot_outside))
```

```python
# Exercise 3: Change a Variable
score = 75
print("original: " + str(score))
score = 90
print("updated:  " + str(score))
```

```python
# Exercise 4: Work With Numbers
number1 = 20
number2 = 5
print("number1 = " + str(number1) + ", " + "number2 = " + str(number2))
print("sum: " + str(number1) + " + " + str(number2) + " = " + str(number1 + number2))
print("difference: " + str(number1) + " - " + str(number2) + " = " + str(number1 - number2))
print("product: " + str(number1) + " * " + str(number2) + " = " + str(number1 * number2))
print("quotient: " + str(number1) + " / " + str(number2) + " = " + str(number1 / number2))
```

```python
# Exercise 5: Use an f-String
first_name = "Richard"
last_name = "Anderson"
interest = "computer programming"
print(f"My name is {first_name} {last_name} and I am interested in {interest}.")
```
