Create a Python program that asks a user to enter personal information and then displays a formatted summary.

Your program must ask the user for:

1. First name
2. Last name
3. Age
4. City
5. Academic major or career interest

Store each answer in an appropriately named variable.

### Program Requirements
Your program must:

Include at least one Python comment.
Use `input()` to collect information.
Store each response in a meaningful variable.
Convert the user's age to an integer using `int()`.
Calculate how old the user will be next year.
Use f-strings to display the results.
Produce a clearly formatted personal information summary.
Run without unexplained errors.
Your program should produce output similar to:


Your program does not have to look exactly like the example, but it must meet all of the requirements.



````python
# Module 2
# Project: Personal Information Form
"""
Program: COP1000_Module02_PersonalInformation.py
Author: Richard Anderson
Project: Personal Information Form
"""
print("== Personal Information Form ==")
# input
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")
career_interest = input("Enter your career interest: ")
# calculations
age_next_year = age + 1
# output
print("== Personal Information Summary ==")
print(f"Name: {first_name} {last_name}")
print(f"Age: {age}")
print(f"Age Next Year: {age_next_year}")
print(f"City: {city}")
print(f"Career Interest: {career_interest}")
````
