# Practice 2: Sets and Collection Operations

```python
# Exercise 1: Create a Set
colors = {"red", "blue", "green", "yellow"}
print(colors)
print(f"number of items in set: {len(colors)}")
```

```python
# Exercise 2: Add and Remove Values
courses = {"COP 1000", "ENC 1101", "MAT 1033"}
print(f"before: {courses}")
courses.add("SLS 1101")
courses.remove("MAT 1033")
print(f" after: {courses}")
```

```python
# Exercise 3: Remove Duplicate Values
names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]
print(f"as a list:\n{names}")
names_set = set(names)
print(f"converted to a set (removes duplicates):\n{names_set}")
print(f"      number of values in list: {len(names)}")
print(f"number of unique values in set: {len(names_set)}")
```

```python
# Exercise 4: Test Membership [test 1]
languages = {"Python", "Java", "C++", "JavaScript"}
user_language = input("Enter a programming language: ")
if user_language in languages:
  print(f"{user_language} is in the set.")
else:
  print(f"{user_language} is not in the set.")
```

```python
# Exercise 5: Find Common Values
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}
common = student_a & student_b
print(f"common: {common}")
```

```python
# Exercise 6: Find All Unique Values
unique = student_a ^ student_b
print(f"unique: {unique}")
```

```python
# Exercise 7: Compare Course Enrollments
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}
in_both = python_students & web_students
in_either = python_students | web_students
python_only = python_students - web_students
print(f"Students in both courses: {in_both}")
print(f"Students in either course: {in_either}")
print(f"Students in Python only: {python_only}")
```
