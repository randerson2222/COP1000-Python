```python
# Exercise 1: Create a Dictionary
student = {
    "name": "Jordan",
    "major": "Cybersecurity",
    "credits": 30
}
print(student)
print(f"   Name: {student["name"]}")
print(f"  Major: {student["major"]}")
print(f"Credits: {student["credits"]}")
```

```python
# Exercise 2: Add a New Key-Value Pair
student["gpa"] = 3.5
print(student)
print(f"GPA: {student["gpa"]}")
```

```python
# Exercise 3: Update a Dictionary Value
course = {
    "course": "COP 1000",
    "room": "A101",
    "students": 24
}
print(course)
course["room"] = "B205"
course["students"] = 26
print(course)
```

```python
# Exercise 4: Remove Data
profile = {
    "name": "Taylor",
    "city": "Jacksonville",
    "age": 21,
    "favorite_color": "blue"
}
print(profile)
print(f"number of key-value pairs remaining using len(): {len(profile)}")
profile.pop("favorite_color")
print(profile)
print(f"number of key-value pairs remaining using len(): {len(profile)}")
```

```python
# Exercise 5: Check for a Key
inventory = {
    "notebooks": 12,
    "pens": 30,
    "folders": 8
}

if "pens" in inventory:
  print("Pens are in the inventory.")
else:
  print("Pens are not in the inventory.")

if "markers" in inventory:
  print("Markers are in the inventory.")
else:
  print("Markers are not in the inventory.")
```

```python
# Exercise 6: Loop Through a Dictionary
prices = {
    "Notebook": 3.49,
    "Pen": 1.25,
    "Folder": 2.00,
    "Backpack": 24.99
}

for key, value in prices.items():
    print(f"{key}: ${value:.2f}")
```

```python
# Exercise 7: Build a Simple Contact Record
contact = {}
contact["Name"] = input("Enter your name: ")
contact["Phone"] = input("Enter your phone number: ")
contact["Email"] = input("Enter your email address: ")
contact["City"] = input("Enter the city in which you reside: ")

print("\n== Contact Information ==")

for key, value in contact.items():
    print(f"{key}: {value}")
```
