# Practice 1: Creating and Calling Functions

```python
# Exercise 1: Create Your First Function
def welcome():
  print("Welcome to COP 1000!")
  print("Python functions help organize programs.")

welcome()
```

```python
# Exercise 2: Call a Function More Than Once
def show_message():
  print("Practice makes progress!")

show_message()
show_message()
show_message()
```

```python
# Exercise 3: Create Multiple Functions
def show_header():
  print("/--------------------\\")
def show_course():
  print("COP 1000\nIntroduction to Python")
def show_footer():
  print("\\--------------------/")

show_header()
show_course()
show_footer()
```

```python
# Exercise 4: Use a Function With a Parameter
def greet(name):
  print(f"Hello, {name}!")

greet("Jordan")
greet("Daniel")
greet("Jacob")
```

```python
# Exercise 5: Use Multiple Parameters
def student_info(name, major):
  print(f"{name} is studying {major}.")

student_info("Taylor", "Cybersecurity")
student_info("Jacob", "Business Administration")
```

```python
# Exercise 6: Pass User Input to a Function
def display_sentence(name, topic):
  print(f"{name}'s favorite Python topic so far is {topic}.")

name = input("Enter your name: ")
topic = input("Enter your favorite topic: ")

display_sentence(name, topic)
```
