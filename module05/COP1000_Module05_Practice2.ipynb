````python
# Exercise 4: Convert Fahrenheit to Celsius
def fahrenheit_to_celsius(fahrenheit):
  return (fahrenheit - 32) * 5 / 9

print("=== Convert Fahrenheit to Celsius ===")
fahrenheit = float(input("Enter temperature in Fahrenheit: "))
celsius = fahrenheit_to_celsius(fahrenheit)
print(f"\nTemperature in Celsius: {celsius:.1f}")
````

## Exercise 6: Separate Calculation From Output

Create two functions:

`def calculate_total(price, quantity):`

and:

`def display_total(item, total):`

The first function should calculate and return the total cost.

The second function should display the item name and total.

Ask the user for:

- Item name
- Price
- Quantity

Your program should use both functions.

Example output:

```
Enter the item: Notebook
Enter the price: 3.50
Enter the quantity: 4

Item: Notebook
Total Cost: $14.00
```

````python
# Exercise 6: Separate Calculation From Output
def calculate_total(price, quantity):
  return price * quantity
def display_total(item, total):
  print(f"Item: {item}")
  print(f"Total Cost: ${total:.2f}")

print("=== Calculate Total Cost (price * quantity) ===")
item_name = input("Enter the item name: ")
price = float(input("Enter the price: "))
quantity = int(input("Enter the quantity: "))
print("")
total = calculate_total(price, quantity)
display_total(item_name, total)
````

## Exercise 7: Design a Small Multi-Function Program
Create a program with at least three functions:

1. A function that displays a title.
2. A function that performs a calculation and returns a value.
3. A function that displays the final result.

You may choose the calculation.

Examples include:

- Miles to kilometers
- Hours to minutes
- Price after discount
- Total cost of several items
- Age in months

Your program should clearly separate each task into a different function.

````python
# Exercise 7: Design a Small Multi-Function Program
def display_title():
  print("=== Discounted Price Calculator ===")
def calculate_discounted_price(price, discount_percent_decimal):
  return (price) - (price * discount_percent_decimal)
def display_result(result):
  print(f"Discounted price: ${result:.2f}")
display_title()
price = float(input("Enter original price: $"))
discount_percent_decimal = float(input("Enter discount percent % (not as decimal): ")) / 100
result = calculate_discounted_price(price, discount_percent_decimal)
print("")
display_result(result)
````
