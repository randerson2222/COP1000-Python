# Practice 1: Operators and Boolean Expressions

````python
# Exercise 1: Arithmetic Operators
number1 = 18
number2 = 5
print("== Python Arithmetic Operators ==")
print(f"Addition: {number1} + {number2} = {number1 + number2}")
print(f"Subtraction: {number1} - {number2} = {number1 - number2}")
print(f"Multiplication: {number1} * {number2} = {number1 * number2}")
print(f"Division: {number1} / {number2} = {number1 / number2}")
print(f"Floor division: {number1} // {number2} = {number1 // number2}")
print(f"Remainder/modulo: {number1} % {number2} = {number1 % number2}")
print(f"Exponentiation: {number1} ** {number2} = {number1 ** number2}")
````

````python
# Exercise 2: Comparison Operators
age = 21
print("== Python Comparison Operators ==")
print(f"Is age equal to 21? {age == 21}")
print(f"Is age not equal to 18? {age != 18}")
print(f"Is age greater than 18? {age > 18}")
print(f"Is age less than 30? {age < 30}")
print(f"Is age greater than or equal to 21? {age >= 21}")
print(f"Is age less than or equal to 20? {age <= 20}")
````

````python
# Exercise 4: Boolean Operators
age = 22
has_id = True
print("== Python Boolean Operators ==")
# and operator
print(f"age >= 18 and has_id => {age >= 18 and has_id}")
# or operator
print(f"age < 18 or has_id => {age < 18 or has_id}")
# not operator
print(f"not has_id => {not has_id}")
````

````python
# Exercise 5: Combine Comparisons
temperature = 75
print("Temperature is...")
print(f"greater than 70? {temperature > 70}")
print(f"less than 90? {temperature < 90}")
print(f"between 70 and 90? {temperature > 70 and temperature < 90}")
print(f"outside the range 60 through 85? {temperature < 60 or temperature > 85}")
````

````python
# Exercise 6: Build a Simple Eligibility Check
age = 19
credits_completed = 30
print("== Student Eligibility Check ==")
print(f"Age Requirement Met: {age >= 18}")
print(f"Credit Requirement Met: {credits_completed >= 24}")
print(f"All Requirements Met: {age >= 18 and credits_completed >= 24}")
````
