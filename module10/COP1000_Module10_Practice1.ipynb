# Practice 1: Creating Classes and Objects

```python
# Exercise 1: Create Your First Class
class Student:
    pass
student1 = Student()
student1.name = "Jordan"
student1.major = "Information Technology"
# Name: Jordan
# Major: Information Technology
print(f"Name: {student1.name}")
print(f"Major: {student1.major}")
```

```python
# Exercise 2: Create Multiple Objects
students = []
for i in range(3):
  students.append(Student())
students[0].name = "Jack"
students[1].name = "James"
students[2].name = "John"
students[0].major = "Computer Science"
students[1].major = "Business"
students[2].major = "Psychology"
students[0].credits = 24
students[1].credits = 20
students[2].credits = 16
for student in students:
  print(f"Name: {student.name}")
  print(f"\tMajor: {student.major}")
  print(f"\tCredits Completed: {student.credits}")
```

```python
# Exercise 3: Create a Class With __init__
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
books = [Book("1984", "George Orwell"), Book("The Great Gatsby", "F. Scott Fitzgerald")]
# books.append(Book("1984", "George Orwell"))
# books.append(Book("The Great Gatsby", "F. Scott Fitzgerald"))
for book in books:
  print(f"Title: {book.title}\n\tAuthor: {book.author}")
```

```python
# Exercise 4: Add More Attributes
class Book:
    def __init__(self, title, author, year, is_available=True):
        self.title = title
        self.author = author
        self.year = year
        self.is_available = is_available
books = [Book("1984", "George Orwell", 1949, True),
         Book("The Great Gatsby", "F. Scott Fitzgerald", 1925, True)]
for book in books:
  print(f"Title: {book.title}")
  print(f"\tAuthor: {book.author}")
  print(f"\tYear: {book.year}")
  print(f"\tAvailable: {book.is_available}")
```

```python
# Exercise 5: Update an Object Attribute
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
product1 = Product("Computer Mouse", 15.00, 10)
print(f"Original Quantity: {product1.quantity}")
product1.quantity -= 3
print(f"Updated Quantity: {product1.quantity}")
```

```python
# Exercise 6: Use Objects in a List
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
book1 = Book("1984", "George Orwell")
book2 = Book("The Great Gatsby", "F. Scott Fitzgerald")
book3 = Book("The Martian", "Andy Weir")
books = [book1, book2, book3]
for book in books:
  print(f"Title: {book.title}\n\tAuthor: {book.author}")
```

```python
# Exercise 7: Create Your Own Class
class Movie:
  def __init__(self,
               title, director,
               distributor, issue_distributor,
               year, issue_year,
               issue_format):
    self.title = title
    self.director = director
    self.distributor = distributor
    self.issue_distributor = issue_distributor
    self.year = year
    self.issue_year = issue_year
    self.issue_format = issue_format
movies = [Movie("2001: A Space Odyssey",
                "Stanley Kubrick",
                "Metro-Goldwyn-Mayer",
                "Warner Bros.",
                1968,
                2018,
                "Ultra HD Blu-ray"),
          Movie("A Christmas Story",
                "Bob Clark",
                "MGM/UA Entertainment Co.",
                "Warner Bros.",
                1983,
                2008,
                "Blu-ray"),
          Movie("The Dark Crystal",
                ["Jim Henson", "Frank Oz"],
                "Associated Film Distribution",
                "Sony Pictures",
                1982,
                2018,
                "Ultra HD Blu-ray")]
for movie in movies:
  print(f"Title: {movie.title}")
  print(f"\tDirector: {movie.director}")
  print(f"\tDistributor: {movie.distributor}")
  print(f"\tIssue Distributor: {movie.issue_distributor}")
  print(f"\tYear: {movie.year}")
  print(f"\tIssue Year: {movie.issue_year}")
  print(f"\tIssue Format: {movie.issue_format}")
```
