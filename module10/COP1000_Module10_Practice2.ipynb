```python
# Exercise 1: Add a Method
class Student:
  def __init__(self, name, major):
    self.name = name
    self.major = major
  def display_info(self):
    print(f"Student: {self.name}")
    print(f"Major: {self.major}")
student1 = Student("Jordan", "Cybersecurity")
student1.display_info()
```

```python
# Exercise 2: Create a Calculation Method
class Rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width
  def calculate_area(self):
    return self.length * self.width
rect1 = Rectangle(10, 2)
rect2 = Rectangle(10, 5)
print(f"rect1 area: {rect1.calculate_area()}")
print(f"rect2 area: {rect2.calculate_area()}")
```

```python
# Exercise 3: Create a Method That Changes an Attribute
class BankAccount:
  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance
  def deposit(self, amount):
    self.balance += amount
account1 = BankAccount("Jordan", 500.00)
print(f"Starting Balance: ${account1.balance:.2f}")
deposit_amount = 100.00
print(f"Deposit: ${deposit_amount:.2f}")
account1.deposit(deposit_amount)
print(f"New Balance: ${account1.balance:.2f}")
```

```python
# Exercise 4: Add Another Behavior
class BankAccount:
  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance
  def deposit(self, amount):
    self.balance += amount
  def withdraw(self, amount):
    if amount > self.balance:
      print("Insufficient funds.")
      return False
    else:
      self.balance -= amount
      return True
account1 = BankAccount("Jordan", 500.00)
print(f"Balance: ${account1.balance:.2f}")
account1.withdraw(400.00)
print(f"Balance: ${account1.balance:.2f}")
account1.withdraw(100.00)
print(f"Balance: ${account1.balance:.2f}")
account1.withdraw(100.00)
print(f"Balance: ${account1.balance:.2f}")
account1.deposit(100.00)
print(f"Balance: ${account1.balance:.2f}")
account1.withdraw(200.00)
print(f"Balance: ${account1.balance:.2f}")

print("")

assert BankAccount("Name", 100.00).withdraw(100.00) == True, "fail"
assert BankAccount("Name", 100.00).withdraw(200.00) == False, "fail"

print("")

acc = BankAccount("Name", 100.00)
acc.withdraw(100.00)
assert acc.balance == 0.00, "fail"
acc.withdraw(100.00)
assert acc.balance == 0.00, "fail"
acc = BankAccount("Name", 100.00)
acc.withdraw(200.00)
assert acc.balance == 100.00, "fail"
```

```python
# Exercise 5: Model a Book
class Book:
  def __init__(self, title, author, availability=True):
    self.title = title
    self.author = author
    self.availability = availability
  def check_out(self):
    if self.availability == False:
      print("book is not available to check out")
    else:
      self.availability = False
  def return_book(self):
    if self.availability == True:
      print("book is not out")
    else:
      self.availability = True
  def print_status(self):
    if self.availability == True:
      print("Status: Available")
    else:
      print("Status: Not Available")
book1 = Book("1984", "George Orwell")
book1.print_status()
book1.return_book()
book1.print_status()
book1.check_out()
book1.print_status()
book1.check_out()
book1.print_status()
book1.return_book()
book1.print_status()
```

```python
# Exercise 6: Multiple Objects With Methods
books = [Book("1984", "George Orwell"),
         Book("The Great Gatsby", "F. Scott Fitzgerald"),
         Book("The Martian", "Andy Weir")]
def show_books(books):
  for book in books:
    print(f"Title: {book.title}")
    print(f"\tAuthor: {book.author}")
    print("\t", end="")
    book.print_status()
show_books(books)
books[0].check_out()
print("")
show_books(books)
########################## alt version ###########################
print("")
book1 = Book("1984", "George Orwell")
book2 = Book("The Great Gatsby", "F. Scott Fitzgerald")
book3 = Book("The Martian", "Andy Weir")
books = [book1, book2, book3]
def show_books(books):
  for book in books:
    print(f"Title: {book.title}")
    print(f"\tAuthor: {book.author}")
    print("\t", end="")
    book.print_status()
show_books(books)
book1.check_out()
book2.check_out()
book3.check_out()
print("")
show_books(books)
```

```python
# Exercise 7: Design Your Own Class
class GameCharacter:
  def __init__(self, name, health=100, skill_level=1):
    self.name = name
    self.health = health
    self.skill_level = skill_level
  def take_damage(self, amount=5):
    self.health = max(0, self.health - amount)
  def level_up(self, amount=1):
    self.skill_level += amount
  def show_status(self):
    print(f"Name: {self.name}")
    print(f"\tHealth: {self.health}")
    print(f"\tSkill Level: {self.skill_level}")
warrior = GameCharacter("Warrior", 100, 5)
warrior.show_status()
warrior.level_up()
warrior.show_status()
warrior.level_up(2)
warrior.show_status()
warrior.take_damage()
warrior.show_status()
warrior.take_damage(50)
warrior.show_status()
warrior.take_damage(50)
warrior.show_status()
```
