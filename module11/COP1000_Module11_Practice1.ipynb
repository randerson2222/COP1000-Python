# Practice 1: Using Python Modules and Libraries

```python
# Exercise 1: Import the math Module
import math

print("The value of math.pi:")
print(f"\t{math.pi}")
print("The square root of 81:")
print(f"\t{math.sqrt(81)}")
print("The value of 5 raised to the power of 2 using math.pow():")
print(f"\t{math.pow(5, 2)}")
```

```python
# Exercise 2: Calculate the Area of a Circle
def circle_area(radius):
  return math.pi * math.pow(radius, 2)

radius = float(input("Enter the radius: "))
area = circle_area(radius)
print(f"Area: {area:.2f}")
```

```python
# Exercise 3: Use the random Module
import random

randint = random.randint(1, 10)
print(randint)
print("")

for i in range(3):
  print(random.randint(1, 10))
```

```python
# Exercise 4: Simulate a Coin Flip
randchoice = random.choice(["Heads", "Tails"])
print(randchoice)
print("")

for i in range(3):
  print(random.choice(["Heads", "Tails"]))
```

```python
# Exercise 5: Simulate a Six-Sided Die
for i in range (1, 6):
  print(f"Roll {i}: {random.randint(1, 6)}")
```

```python
# Exercise 6: Import Specific Functions
from math import sqrt

print(f"sqrt(25): {sqrt(25)}")
print(f"sqrt(100): {sqrt(100)}")
print(f"sqrt(144): {sqrt(144)}")
```

#### Exercise 6 answer:

`import math` imports the entire math library and functions have to be accessed like `math.sqrt(25)`.

`from math import sqrt` imports only the sqrt function and that function can be accessed directly like `sqrt(25)`.
