# Practice 2: Working With CSV Data

```python
# Exercise 1: Import the CSV Module
import csv

print("CSV tools are ready.")
```

```python
# Exercise 2: Create a CSV File
header = ["Name", "Major", "Credits"]
rows = [
    ["Alex", "Cybersecurity", "24"],
    ["Jordan", "Information Technology", "36"],
    ["Taylor", "Computer Science", "42"]
]
with open("students.csv", mode="w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(header)
  writer.writerows(rows)
```

```python
# Exercise 3: Read a CSV File
with open("students.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

```python
# Exercise 4: Display Selected Data
with open("students.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  # Skip the header row, return None if file empty
  next(reader, None)
  # Loop through the remaining rows and display Name and Major
  for row in reader:
    name = row[0]
    major = row[1]
    print(f"{name} - {major}")
```

```python
# Exercise 5: Add a New Record
name = input("enter student name: ").strip()
major = input("enter major: ").strip()
credits = input("enter number of completed credits: ").strip()
row = [name, major, credits]
with open("students.csv", mode="a", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(row)
with open("students.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

```python
# Exercise 6: Calculate From CSV Data
total_credits = 0
student_count = 0

with open("students.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)

  # Skip the header row
  next(reader, None)

  for row in reader:
    credits = int(row[2])
    total_credits += credits
    student_count += 1

# Avoid division by zero if the file has no student rows
average_credits = total_credits / student_count if student_count > 0 else 0

print(f"Student Records: {student_count}")
print(f"Total Credits: {total_credits}")
print(f"Average Credits: {average_credits:.2f}")
```

```python
# Exercise 7: Create Your Own CSV Dataset
header = ["Product", "Price", "Quantity"]
rows = [
    ["Notebook", "3.49", "4"],
    ["Pen", "1.25", "10"],
    ["Folder", "2.00", "5"]
]
with open("products.csv", mode="w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(header)
  writer.writerows(rows)
with open("products.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  next(reader, None)
  for row in reader:
    print(f"Product: {row[0]}")
    print(f"\tPrice: ${row[1]}")
    print(f"\tQuantity: ${row[2]}")
```
