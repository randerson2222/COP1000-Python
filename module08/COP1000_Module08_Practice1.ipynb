# Practice 1: Strings and Text File I/O

```python
# Exercise 1: Work With Strings
message = "Python programming is useful"
print(message)
print(len(message))
print(message.upper())
print(message.lower())
updated = message.replace("useful", "powerful")
print(updated)
```

```python
# Exercise 2: String Indexing and Slicing
course = "COP1000 Python Programming"
print(course[0])
print(course[len(course)-1])
print(course[-1])
print(course[0:7])
print(course[8:14])
```

```python
# Exercise 3: Clean User Input
name = input("enter your name with extra spaces before and/or after it:\n").strip()
print(f"Hello, {name}!")
```

```python
# Exercise 4: Write to a Text File
with open("student.txt", "w") as file:
  file.write("Richard Anderson\n")
  file.write("COP 1000\n")
  file.write("Learning python is rewarding.\n")
print("File created successfully.")
```

```python
# Exercise 5: Read From a Text File
with open("student.txt", "r") as file:
  print(file.read())
```

```python
# Exercise 6: Append to a File
with open("student.txt", "a") as file:
  file.write("Module 8: File I/O\n")
with open("student.txt", "r") as file:
  print(file.read())
```

```python
# Exercise 7: Create a Simple Notes File
with open("notes.txt", "w") as file:
  for i in range(1,4):
    file.write(input(f"Enter note {1}: ").strip() + "\n")
with open("notes.txt", "r") as file:
  print("Saved Notes: (whole file)")
  print(file.read())
with open("notes.txt", "r") as file:
  print("Saved Notes: (loop over lines)")
  for line in file:
    clean_line = line.rstrip("\n")
    print(clean_line)
```
