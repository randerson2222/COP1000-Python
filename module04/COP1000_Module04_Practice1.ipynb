# Practice 1: `for` Loops and `range()`

```python
# Exercise 1: Repeat a Message
for i in range(5):
  print("Python makes repetition easier!")
```

```python
# Exercise 2: Count From 1 to 10
for i in range(1, 11):
  print(i)
```

```python
# Exercise 3: Count by Twos
for i in range(2, 21, 2):
  print(i)
```

```python
# Exercise 4: Count Backward
for i in range(10, 0, -1):
    print(i)
print("Blast off!")
```

```python
# Exercise 5: Create a Multiplication Table
number = int(input("Enter a number: "))
print("")
for i in range(1, 11):
  print(f"{number} x {i} = {number * i}")
```

```python
# Exercise 6: Calculate a Running Total
total = 0
for i in range(1, 11):
  total += i
print(f"The total is {total}.")
```

```python
# Exercise 7: Calculate Five Test Scores
total = 0
for i in range(1, 6):
  test_score = float(input(f"Enter test score {i}: "))
  total += test_score
print("")
print(f"Total: {total:.2f}")
average = total / 5
print(f"Average: {average:.2f}")
```
