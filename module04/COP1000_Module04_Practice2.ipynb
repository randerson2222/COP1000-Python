# Practice 2: `while` Loops and Loop Control

```python
# Exercise 1: Count With a while Loop
number = 1
while (number <= 10):
  print(number)
  number += 1
```

```python
# Exercise 2: Countdown
count = 5
while (count >= 1):
  print(count)
  count -= 1
print("Done!")
```

```python
# Exercise 3: Repeat Until the User Stops
while True:
  user_input = input("Would you like to continue? Enter yes or no: ")
  if user_input == "no":
    break
  # lines below within while block can be removed for equivalent functionality
  elif user_input == "yes":
    pass
  else:
    pass
print("Program finished.")
```

```python
# Exercise 4: Password Check
password = "python123"
while True:
  user_input = input("Enter the password: ")
  if user_input == password:
    print("Access granted.")
    break
  else:
    print("Incorrect password. Try again.")
```

```python
# Exercise 5: Validate Numeric Input
range_min = 1
range_max = 10
while True:
  user_input = input(f"Enter a number from {range_min} to {range_max}: ")
  try:
    user_number = int(user_input)
    if user_number >= range_min and user_number <= range_max:
      print(f"You entered a valid number: {user_number}")
      break
    else:
      print(f"Invalid number. Try again. (number not in range {range_min} to {range_max})")
  except ValueError:
    print("Invalid number. Try again. (input cannot be converted to integer)")
```

```python
# Exercise 5: Validate Numeric Input [alt structure]
range_min = 1
range_max = 10
while True:
  user_input = input(f"Enter a number from {range_min} to {range_max}: ")
  try:
    user_number = int(user_input)
  except ValueError:
    print("Invalid number. Try again. (input cannot be converted to integer)")
    # continue jumps back to conditional expression then beginning of interior of loop if conditional remains true
    continue
  if user_number >= range_min and user_number <= range_max:
    print(f"You entered a valid number: {user_number}")
    break
  else:
    print(f"Invalid number. Try again. (number not in range {range_min} to {range_max})")
```

```python
# Exercise 6: Running Total Until Zero
total = 0
stop_val = 0
while True:
  user_input = input(f"Enter a number or {stop_val} to stop: ")
  try:
    user_number = int(user_input)
    if user_number == stop_val:
      break
    else:
      total += user_number
  except ValueError:
    print("Please enter a number. (input cannot be converted to integer)")
print(f"\nTotal: {total}")
```

```python
# Exercise 7: Limited Attempts [test 1]
word = "python"
max_attempts = 3
print(f"You get {max_attempts} attempts to enter the correct word.")

counter = 0
while True:
  if counter >= max_attempts:
    print("No attempts remaining.")
    break
  user_input = input("Enter the correct word: ")
  if user_input == word:
    print("Correct!")
    break
  else:
    counter += 1
    print(f"Incorrect. You used attempt {counter}")
```
