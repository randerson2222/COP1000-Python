# Practice 2: Making API Requests and Processing External Data

```python
# Exercise 1: Import requests
import requests

print("Requests library loaded.")
```

```python
# Exercise 2: Make Your First API Request
url = "https://geocoding-api.open-meteo.com/v1/search"
params = {
    "name": "Jacksonville",
    "count": 1,
    "language": "en",
    "format": "json"
}
response = requests.get(url, params=params, timeout=15)
print(response.url)
print(f"Status Code: {response.status_code}")
```

```python
# Exercise 3: Convert the Response to Python Data
from pprint import pprint

data = response.json()
pprint(data)
print(type(data))
```

```python
# Exercise 4: Extract Location Information
location = data["results"][0]
print(f"Location: {location["name"]}")
print(f"State: {location["admin1"]}")
print(f"Country: {location["country"]}")
print(f"Latitude: {location["latitude"]}")
print(f"Longitude: {location["longitude"]}")
```

```python
# Exercise 5: Make a Weather Request
latitude = location["latitude"]
longitude = location["longitude"]
weather_url = "https://api.open-meteo.com/v1/forecast"
weather_params = {
    "latitude": latitude,
    "longitude": longitude,
    "current": "temperature_2m,wind_speed_10m",
    "temperature_unit": "fahrenheit",
    "wind_speed_unit": "mph"
}
weather_response = requests.get(weather_url, params=weather_params, timeout=15)
print(weather_response.url)
print(f"Status Code: {weather_response.status_code}")
weather_data = weather_response.json()
pprint(weather_data)
print(type(weather_data))
```

```python
# Exercise 6: Extract Current Weather
print("== Current Weather ==")
temp_key = "temperature_2m"
wind_speed_key = "wind_speed_10m"
print(f"Temperature {weather_data["current"][temp_key]} {weather_data["current_units"][temp_key]}")
print(f"Wind Speed {weather_data["current"][wind_speed_key]} {weather_data["current_units"][wind_speed_key]}")
```
