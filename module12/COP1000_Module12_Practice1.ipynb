 # Practice 1: Working With JSON Data

```python
# Exercise 1: Review Dictionary-Style Data
student = {
    "name": "Jordan",
    "major": "Information Technology",
    "credits": 36
}
print(f"Name: {student["name"]}")
print(f"Major: {student["major"]}")
print(f"Credits: {student["credits"]}")
```

```python
# Exercise 2: Work With Nested Data
weather = {
    "city": "Jacksonville",
    "current": {
        "temperature": 82,
        "humidity": 68,
        "wind_speed": 9
    }
}
print(f"City: {weather["city"]}")
print(f"Temperature: {weather["current"]["temperature"]}")
print(f"Humidity: {weather["current"]["humidity"]}")
print(f"Wind Speed: {weather["current"]["wind_speed"]}")
```

```python
# Exercise 3: Work With a List Inside a Dictionary
forecast = {
    "city": "Jacksonville",
    "temperatures": [82, 84, 86, 83, 79]
}
print(f"complete list: {forecast["temperatures"]}")
print(f"  first temp: {forecast["temperatures"][0]}")
print(f"   last temp: {forecast["temperatures"][-1]}")
print(f"highest temp: {max(forecast["temperatures"])}")
print(f" lowest temp: {min(forecast["temperatures"])}")
```

```python
# Exercise 4: Process Multiple Records
data = {
    "cities": [
        {"name": "Jacksonville", "temperature": 82},
        {"name": "Orlando", "temperature": 87},
        {"name": "Tallahassee", "temperature": 84}
    ]
}
for city in data["cities"]:
  print(f"{city["name"]}: {city["temperature"]}")
```

```python
# Exercise 5: Convert JSON Text to Python Data
import json

json_text = '{"name": "Alex", "age": 22, "major": "Computer Science"}'

py_dict = json.loads(json_text)

print(f"Name: {py_dict["name"]}")
print(f"Age: {py_dict["age"]}")
print(f"Major: {py_dict["major"]}")
print(f"\nPython Type: {type(py_dict).__name__}")
```

```python
# Exercise 6: Convert Python Data to JSON
course = {
    "course": "COP 1000",
    "language": "Python",
    "active": True
}

json_text = json.dumps(course)
print(json_text)
print(f"\nPython Type: {type(json_text).__name__}")
```
