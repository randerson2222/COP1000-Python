```python
# Exercise 1: Loop Through a List
animals = ["dog", "cat", "rabbit", "horse", "turtle"]
for animal in animals:
  print(animal)
```

```python
# Exercise 2: Number the Items
tasks = ["Study Python", "Complete homework", "Read chapter", "Review notes"]
for index, task in enumerate(tasks, start=1):
  print(f"{index}. {task}")
```

```python
# Exercise 4: Filter Values
temperatures = [45, 72, 88, 61, 93, 55, 79]
count = 0
for temperature in temperatures:
  if temperature >= 70:
    print(temperature)
    count += 1
print(f"{count} temperatures met the requirement.")
```

```python
# Exercise 5: Calculate an Average From User Data
scores = []
for i in range(1,6):
  scores.append(float(input(f"Enter test score #{i}: ")))
print(f" scores: {scores}")
total = sum(scores)
print(f"  total: {total}")
average = total / len(scores)
print(f"average: {average:.2f}")
```

```python
# Exercise 6: Create a Tuple
weekend = ("Saturday", "Sunday")
print(weekend)
print(weekend[0])
print(weekend[1])
print(len(weekend))
for day in weekend:
  print(day)
```
