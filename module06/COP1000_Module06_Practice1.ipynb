# Practice 1: Creating and Modifying Lists

```python
# Exercise 1: Create a List
foods = ["pizza", "tacos", "pasta", "salad", "burgers"]
print(foods[0])
print(foods[2])
print(foods[-1])
```

```python
# Exercise 2: Change an Item
colors = ["red", "blue", "green", "yellow"]
print(f"before: {colors}")
colors[colors.index("green")] = "purple"
print(f" after: {colors}")
```

```python
# Exercise 3: Add Items to a List
courses = ["COP 1000", "ENC 1101"]
courses.append("MAC 1000")
courses.insert(0, "PSY 1000")
print(courses)
```

```python
# Exercise 4: Remove Items
shopping = ["milk", "bread", "eggs", "apples", "coffee"]
shopping.remove("eggs")
shopping.pop(-1)
print(shopping)
print(f"number of items remaining: {len(shopping)}")
```

```python
# Exercise 5: Sort a List
names = ["Andy", "Colin", "Dave", "Terry", "Barry"]
print(f"original: {names}")
names.sort()
print(f"  sorted: {names}")
names.reverse()
print(f"reversed: {names}")
```

```python
# Exercise 6: Work With Numbers in a List
scores = [88, 94, 76, 91, 85]
print(f"   highest score: {max(scores)}")
print(f"    lowest score: {min(scores)}")
total = sum(scores)
number_of_scores = len(scores)
average = total / number_of_scores
print(f"           total: {total}")
print(f"number of scores: {number_of_scores}")
print(f"         average: {average:.2f}")
```

```python
# Exercise 7: Build a Simple Student List
students = []
for i in range(1,4):
  students.append(input(f"Enter student name #{i}: "))
print(f"students: {students}")
print(f"number of students: {len(students)}")
students.sort()
print(f"sorted students: {students}")
```
