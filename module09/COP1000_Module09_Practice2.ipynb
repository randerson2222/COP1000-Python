```python
# Exercise 3: Identify a Runtime Error [no exception handling]
number = 10
result = number / 0
print(result)
```

```python
# Exercise 3: Identify a Runtime Error [with exception handling]
number = 10
try:
  result = number / 0
  print(result)
except ZeroDivisionError:
  print("Division by zero is not allowed.")
```

```python
# Exercise 4: Find a Logic Error [broken]
length = 10
width = 5

area = length + width

print(f"Area: {area}")
```

```python
# Exercise 4: Find a Logic Error [fix]
length = 10
width = 5

area = length * width

print(f"Area: {area}")
```

#### Exercise 4 answer:

This is a logic error rather than a syntax error or runtime error because the program is well-formed and runs without errors, but produces the incorrect result, because incorrect logic was implemented. In this case, the formula for area is length times width, requiring the multiplication operator. In the broken example, the addition operator was used instead. The program runs fine, but produces the wrong result.

```python
# Exercise 6: Debug a Loop [broken - runs forever]
count = 1

while count <= 5:
    print(count)
```

```python
# Exercise 6: Debug a Loop [fix]
count = 1

while count <= 5:
    print(count)
    count += 1
```

#### Exercise 6 answer:

In the broken example, the statement to increment the count variable was not added, so the count did not increase on each loop iteration. In addition, the conditional in the while loop always remained true, causing the loop to run forever.

```python
# Exercise 7: Test a Program With Multiple Values
test_data = [
    [100, 20, None],
    [50, 10, None],
    [25, 0, None]
]
test_data_expected_out = [
    [100, 20, 80.0],
    [50, 10, 45.0],
    [25, 0, 25.0]
]

def calculate_discount(price, discount_percent):
  return price - (price * (discount_percent / 100))

# assert calculate_discount(100, 20) == 80.0, "fail"
# assert calculate_discount(50, 10) == 45.0, "fail"
# assert calculate_discount(25, 0) == 25.0, "fail"

for record in test_data:
  record[2] = calculate_discount(record[0], record[1])

print(f"expected: {test_data_expected_out}")
print(f"  result: {test_data}")

assert test_data == test_data_expected_out, "unexpected out"

print("tests passed")
```

#### Exercise 7 answer:

Each test produced the expected result, matching the table in the practice instructions, which was converted to a list of lists in Python as `test_data_expected_out`.
