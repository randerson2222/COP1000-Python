```python
# Exercise 3: Protect a Division Calculation [test 1 - valid]
try:
  first = int(input("Enter the first number: ").strip())
  second = int(input("Enter the second number: ").strip())
except ValueError:
  print("Please enter a valid whole number.")

try:
  quotient = first / second
  print(f"{first} / {second} = {quotient}")
except ZeroDivisionError:
  print("You cannot divide by zero.")
```

```python
# Exercise 4: Validate a Grade [test 1 - valid]
grade = None
try:
  input_int = int(input("Enter a grade: ").strip())
  if 0 <= input_int <= 100:
    grade = input_int
  else:
    print("Grade must be between 0 and 100.")
except ValueError:
  print("Please enter a valid whole number.")
print(f"Grade: {grade}")
```

```python
# Exercise 5: Keep Asking Until Input Is Valid
age = None
while age == None:
  try:
    input_int = int(input("Enter your age: ").strip())
    if input_int > 0:
      age = input_int
    else:
      print("Age must be greater than zero.")
  except ValueError:
    print("Invalid input. Enter a whole number.")
print(f"Age accepted: {age}")
```

```python
# Exercise 7: Safe Average Calculator [test 1 - valid]
scores = []
try:
  for i in range(1,4):
    scores.append(int(input(f"Enter score {i}: ").strip()))
  average = sum(scores) / len(scores)
  print(f"Average: {average:.2f}")
except ValueError:
  print("Invalid input. Enter a whole number.")
```
